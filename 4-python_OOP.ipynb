# Часть 1: Зачем нужно ООП? (10 минут)
## 1.1. От процедурного к объектному

# Почему ООП важно для ML?

Представьте, что мы хотим работать с данными о студентах.

**Процедурный подход:**
- Данные хранятся отдельно (словари, списки)
- Функции обрабатывают эти данные отдельно
- Если структура данных меняется, нужно менять все функции

**Объектно-ориентированный подход:**
- Данные и методы их обработки объединены в одном месте (классе)
- Легко создавать множество похожих объектов
- Удобно расширять функциональность
- **Именно так устроены все ML библиотеки!**

# 1.2. Сравнение подходов на практике

```python
# ПРОЦЕДУРНЫЙ ПОДХОД
# Данные хранятся отдельно
student1 = {"name": "Алиса", "grades": [85, 90, 78], "group": "ML-101"}
student2 = {"name": "Борис", "grades": [92, 88, 95], "group": "ML-101"}

# Функции для работы с данными - отдельно
def calculate_average(student):
    """Вычисляет средний балл студента"""
    return sum(student["grades"]) / len(student["grades"])

def print_student_info(student):
    """Выводит информацию о студенте"""
    avg = calculate_average(student)
    print(f"Студент: {student['name']}, Группа: {student['group']}, Средний балл: {avg:.2f}")

# Использование
print_student_info(student1)
print_student_info(student2)

print("\nПроблема: данные и функции не связаны явно.")
print("Если мы изменим структуру данных, придется менять все функции!")
```

```python
# ОБЪЕКТНО-ОРИЕНТИРОВАННЫЙ ПОДХОД (предварительный просмотр)
# Определим класс (будет подробно объяснено позже)
class Student:
    def __init__(self, name, grades, group):
        self.name = name
        self.grades = grades
        self.group = group
    
    def calculate_average(self):
        return sum(self.grades) / len(self.grades)
    
    def print_info(self):
        avg = self.calculate_average()
        print(f"Студент: {self.name}, Группа: {self.group}, Средний балл: {avg:.2f}")

# Создание объектов
student1_obj = Student("Алиса", [85, 90, 78], "ML-101")
student2_obj = Student("Борис", [92, 88, 95], "ML-101")

# Использование
student1_obj.print_info()
student2_obj.print_info()

print("\nПреимущество: данные и методы их обработки - вместе!")
print("student1_obj.print_info() - метод сам знает, как работать со своими данными.")
```

# Часть 2: Классы и объекты (20 минут)
## 2.1. Создание простого класса

```python
# Определяем класс (шаблон)
class Cat:
    """Простой класс, представляющий кошку."""
    
    # Метод __init__ - конструктор, вызывается при создании объекта
    def __init__(self, name, age):
        """Инициализация атрибутов кошки."""
        self.name = name  # Атрибут объекта
        self.age = age    # Атрибут объекта
        print(f"Создана кошка {self.name}!")
    
    # Метод - функция, принадлежащая классу
    def meow(self):
        """Кошка мяукает."""
        return f"{self.name} говорит: Мяу!"
    
    def celebrate_birthday(self):
        """Увеличиваем возраст на 1."""
        self.age += 1
        print(f"{self.name} теперь {self.age} лет!")

# Создаем объекты (экземпляры класса)
my_cat = Cat("Мурка", 3)
neighbors_cat = Cat("Барсик", 5)

print("\n--- Работа с объектами ---")
print(my_cat.meow())
print(neighbors_cat.meow())

print(f"\nВозраст Мурки: {my_cat.age}")
my_cat.celebrate_birthday()
print(f"После дня рождения: {my_cat.age}")
```

# 2.2. Атрибуты класса и атрибуты объекта

```python
class Dog:
    # Атрибут класса - общий для всех объектов
    species = "Canis familiaris"
    total_dogs = 0  # Счетчик всех созданных собак
    
    def __init__(self, name, breed):
        # Атрибуты объекта - уникальны для каждого
        self.name = name
        self.breed = breed
        Dog.total_dogs += 1  # Обращаемся к атрибуту класса
        print(f"Собака {self.name} создана. Всего собак: {Dog.total_dogs}")
    
    def bark(self):
        return f"{self.name} говорит Гав!"
    
    @classmethod
    def get_total_dogs(cls):
        """Метод класса - работает с классом, а не с объектом."""
        return f"Всего создано собак: {cls.total_dogs}"

# Создаем объекты
dog1 = Dog("Рекс", "Овчарка")
dog2 = Dog("Бим", "Такса")
dog3 = Dog("Лорд", "Доберман")

print("\n--- Атрибуты класса ---")
print(f"Вид всех собак: {Dog.species}")  # Доступ через класс
print(f"Вид dog1: {dog1.species}")       # Доступ через объект
print(Dog.get_total_dogs())

print("\n--- Атрибуты объекта ---")
print(f"dog1.name: {dog1.name}")
print(f"dog2.name: {dog2.name}")

# Изменение атрибута класса
Dog.species = "Canis lupus familiaris"
print(f"\nПосле изменения: {dog1.species} (изменилось у всех!)")
```

# 2.3. Магические методы (str, repr)

Эти методы делают наши объекты более "питоничными".

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):
        """Для пользователей (print, str()). Должен быть читаемым."""
        return f"'{self.title}' от {self.author}"
    
    def __repr__(self):
        """Для разработчиков (отладка, repr()). Должен однозначно идентифицировать объект."""
        return f"Book('{self.title}', '{self.author}', {self.pages})"
    
    def __len__(self):
        """Позволяет использовать len() на объекте."""
        return self.pages
    
    def __add__(self, other):
        """Позволяет складывать книги (возвращает количество страниц)."""
        if isinstance(other, Book):
            return self.pages + other.pages
        return NotImplemented

# Демонстрация
book1 = Book("1984", "Джордж Оруэлл", 328)
book2 = Book("Мастер и Маргарита", "Михаил Булгаков", 480)

print("--- Магические методы ---")
print(str(book1))        # Вызов __str__
print(repr(book1))       # Вызов __repr__
print(book1)             # print использует __str__

print(f"\nДлина книги: {len(book1)} страниц")  # Вызов __len__
print(f"Сумма страниц: {book1 + book2}")       # Вызов __add__
```

# Часть 3: Инкапсуляция и свойства (15 минут)
## 3.1. Публичные и "приватные" атрибуты

В Python нет строгих модификаторов доступа, но есть соглашения.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner                # Публичный атрибут
        self._balance = balance           # "Защищенный" (по соглашению)
        self.__secret_code = 12345        # "Приватный" (name mangling)
    
    def deposit(self, amount):
        """Публичный метод для внесения денег."""
        if amount > 0:
            self._balance += amount
            print(f"Внесено {amount}. Баланс: {self._balance}")
        else:
            print("Сумма должна быть положительной")
    
    def withdraw(self, amount):
        """Публичный метод для снятия денег."""
        if 0 < amount <= self._balance:
            self._balance -= amount
            print(f"Снято {amount}. Баланс: {self._balance}")
        else:
            print("Недостаточно средств или неверная сумма")
    
    def get_balance(self):
        """Геттер для баланса."""
        return self._balance
    
    def __secret_method(self):
        """Приватный метод."""
        return "Это секретный метод"

# Демонстрация
account = BankAccount("Алиса", 1000)

print("--- Работа со счетом ---")
print(f"Владелец: {account.owner}")           # Доступно
print(f"Баланс через геттер: {account.get_balance()}")

account.deposit(500)
account.withdraw(200)

# Доступ к "защищенному" атрибуту (можно, но не рекомендуется)
print(f"Баланс напрямую: {account._balance}")

# Доступ к "приватному" атрибуту
# print(account.__secret_code)  # Ошибка! AttributeError
print(f"Доступ к приватному атрибуту: {account._BankAccount__secret_code}")  # Name mangling
```

## 3.2. Свойства (property) - питонический способ

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius
    
    @property
    def celsius(self):
        """Геттер для celsius."""
        print("Получаем значение celsius")
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        """Сеттер для celsius с валидацией."""
        print("Устанавливаем значение celsius")
        if value < -273.15:
            raise ValueError("Температура ниже абсолютного нуля невозможна")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        """Вычисляемое свойство (только для чтения)."""
        return (self.celsius * 9/5) + 32
    
    @fahrenheit.setter
    def fahrenheit(self, value):
        """Сеттер для fahrenheit (конвертирует в celsius)."""
        self.celsius = (value - 32) * 5/9

# Демонстрация
temp = Temperature(25)
print(f"Температура: {temp.celsius}°C")
print(f"По Фаренгейту: {temp.fahrenheit}°F")

temp.celsius = 30  # Используем сеттер
print(f"Новая температура: {temp.celsius}°C")
print(f"По Фаренгейту: {temp.fahrenheit}°F")

temp.fahrenheit = 100  # Устанавливаем через Фаренгейт
print(f"После установки 100°F: {temp.celsius}°C")

# temp.celsius = -300  # Ошибка! ValueError
```

# Часть 4: Наследование (20 минут)
## 4.1. Базовое наследование

```python
class Vehicle:
    """Базовый класс для всех транспортных средств."""
    
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
        self._mileage = 0
    
    def start(self):
        return f"{self.brand} {self.model} заводится..."
    
    def stop(self):
        return f"{self.brand} {self.model} останавливается."
    
    def drive(self, distance):
        self._mileage += distance
        return f"Проехали {distance} км. Пробег: {self._mileage} км"
    
    def info(self):
        return f"{self.year} {self.brand} {self.model}, пробег: {self._mileage} км"

class Car(Vehicle):
    """Класс автомобиля, наследует от Vehicle."""
    
    def __init__(self, brand, model, year, fuel_type):
        # Вызов конструктора родителя
        super().__init__(brand, model, year)
        self.fuel_type = fuel_type
        self._fuel_level = 100
    
    def refuel(self, amount):
        """Новый метод, специфичный для Car."""
        self._fuel_level = min(100, self._fuel_level + amount)
        return f"Заправлено. Уровень топлива: {self._fuel_level}%"
    
    # Переопределение метода родителя
    def info(self):
        """Расширяем метод родителя."""
        base_info = super().info()
        return f"{base_info}, топливо: {self.fuel_type}, уровень: {self._fuel_level}%"

class ElectricCar(Car):
    """Электромобиль - наследует от Car."""
    
    def __init__(self, brand, model, year, battery_capacity):
        # Для электромобиля fuel_type всегда "electric"
        super().__init__(brand, model, year, "electric")
        self.battery_capacity = battery_capacity
        self._charge_level = 100
    
    def charge(self, amount):
        """Переопределяем refuel для электромобиля."""
        self._charge_level = min(100, self._charge_level + amount)
        return f"Заряжено. Уровень заряда: {self._charge_level}%"
    
    def refuel(self, amount):
        """Можно вызвать charge вместо refuel."""
        return self.charge(amount)
    
    # Полное переопределение метода
    def info(self):
        return f"🔋 Электромобиль: {self.brand} {self.model} ({self.year}), заряд: {self._charge_level}%, батарея: {self.battery_capacity} кВт·ч"

# Демонстрация
print("--- Обычный автомобиль ---")
my_car = Car("Toyota", "Camry", 2022, "бензин")
print(my_car.start())
print(my_car.drive(150))
print(my_car.refuel(30))
print(my_car.info())

print("\n--- Электромобиль ---")
tesla = ElectricCar("Tesla", "Model 3", 2023, 75)
print(tesla.start())
print(tesla.drive(200))
print(tesla.charge(50))
print(tesla.refuel(20))  # Работает через charge
print(tesla.info())

# Проверка типов
print(f"\nmy_car является Car? {isinstance(my_car, Car)}")
print(f"tesla является Vehicle? {isinstance(tesla, Vehicle)}")
print(f"Car является подклассом Vehicle? {issubclass(Car, Vehicle)}")
```

## 4.2. Множественное наследование (кратко)

```python
class Flyable:
    """Примесь (mixin) для летающих объектов."""
    
    def fly(self):
        return "Летит по воздуху"
    
    def land(self):
        return "Приземляется"

class Swimmable:
    """Примесь для плавающих объектов."""
    
    def swim(self):
        return "Плывет по воде"
    
    def dive(self):
        return "Ныряет"

class Duck(Flyable, Swimmable):
    """Утка умеет и летать, и плавать."""
    
    def __init__(self, name):
        self.name = name
    
    def quack(self):
        return f"{self.name} говорит: Кря!"

# Демонстрация
donald = Duck("Дональд")
print(donald.quack())
print(donald.fly())
print(donald.swim())
print(donald.dive())

# Порядок разрешения методов (MRO)
print(f"\nПорядок разрешения методов для Duck:")
for cls in Duck.__mro__:
    print(f"  {cls.__name__}")
```

# Часть 5: Полиморфизм (15 минут)
## 5.1. Полиморфизм в действии

```python
class Animal:
    """Базовый класс."""
    
    def __init__(self, name):
        self.name = name
    
    def make_sound(self):
        """Базовый метод, будет переопределен."""
        return "Издает звук"
    
    def move(self):
        return "Двигается"

class Dog(Animal):
    def make_sound(self):
        return f"{self.name} лает: Гав-гав!"
    
    def move(self):
        return f"{self.name} бегает на четырех лапах"

class Cat(Animal):
    def make_sound(self):
        return f"{self.name} мяукает: Мяу-мяу!"
    
    def move(self):
        return f"{self.name} крадется бесшумно"

class Bird(Animal):
    def make_sound(self):
        return f"{self.name} поет: Чирик-чирик!"
    
    def move(self):
        return f"{self.name} летает по воздуху"
    
    def fly(self):
        return f"{self.name} парит в небе"

# Функция, демонстрирующая полиморфизм
def animal_show(animals):
    """Принимает список любых животных и вызывает их методы."""
    print("--- Зоошоу начинается! ---")
    for animal in animals:
        print(animal.make_sound())
        print(animal.move())
        if hasattr(animal, 'fly'):  # Проверяем наличие метода
            print(animal.fly())
        print("-" * 30)

# Создаем разных животных
animals = [
    Dog("Рекс"),
    Cat("Мурка"),
    Bird("Кеша"),
    Dog("Бобик")
]

# Один и тот же код работает с разными типами!
animal_show(animals)

print("\nПолиморфизм: один интерфейс - разная реализация!")
```

# 5.2. Практический пример для ML

Показываем, как полиморфизм используется в ML библиотеках.

```python
from abc import ABC, abstractmethod
import numpy as np

# Абстрактный базовый класс для моделей ML
class BaseModel(ABC):
    """Абстрактный класс для всех ML моделей."""
    
    def __init__(self, name):
        self.name = name
        self.is_trained = False
    
    @abstractmethod
    def train(self, X, y):
        """Абстрактный метод - должен быть реализован в наследниках."""
        pass
    
    @abstractmethod
    def predict(self, X):
        """Абстрактный метод - должен быть реализован в наследниках."""
        pass
    
    def get_info(self):
        """Общий для всех метод."""
        status = "обучена" if self.is_trained else "не обучена"
        return f"Модель {self.name}: {status}"

# Конкретная реализация - линейная регрессия
class LinearRegression(BaseModel):
    def __init__(self):
        super().__init__("Линейная регрессия")
        self.weights = None
    
    def train(self, X, y):
        print(f"Обучение {self.name}...")
        # Упрощенная имитация обучения
        self.weights = np.random.randn(X.shape[1]) if X.ndim > 1 else np.random.randn(1)
        self.is_trained = True
        return "Модель обучена"
    
    def predict(self, X):
        if not self.is_trained:
            raise ValueError("Модель не обучена!")
        # Имитация предсказания
        return np.random.randn(len(X)) if hasattr(X, '__len__') else np.random.randn(1)

# Конкретная реализация - случайный лес
class RandomForest(BaseModel):
    def __init__(self, n_trees=100):
        super().__init__(f"Случайный лес ({n_trees} деревьев)")
        self.n_trees = n_trees
        self.trees = None
    
    def train(self, X, y):
        print(f"Обучение {self.name}...")
        # Имитация обучения
        self.trees = [f"Tree_{i}" for i in range(self.n_trees)]
        self.is_trained = True
        return f"Обучено {self.n_trees} деревьев"
    
    def predict(self, X):
        if not self.is_trained:
            raise ValueError("Модель не обучена!")
        # Имитация предсказания
        return np.random.randn(len(X)) if hasattr(X, '__len__') else np.random.randn(1)

# Функция для работы с любой моделью
def evaluate_model(model, X_train, y_train, X_test):
    """Работает с любой моделью, наследующей BaseModel."""
    print(f"\n{'='*50}")
    print(model.get_info())
    
    # Обучение
    model.train(X_train, y_train)
    print(model.get_info())
    
    # Предсказание
    predictions = model.predict(X_test)
    print(f"Предсказания для {len(X_test)} samples: {predictions[:5]}...")
    
    return predictions

# Демонстрация
np.random.seed(42)
X_train = np.random.randn(100, 5)
y_train = np.random.randn(100)
X_test = np.random.randn(20, 5)

# Создаем разные модели
models = [
    LinearRegression(),
    RandomForest(50),
    RandomForest(200)
]

# Один и тот же код работает для всех моделей!
for model in models:
    evaluate_model(model, X_train, y_train, X_test)

print("\n" + "!"*50)
print("Это именно так работают sklearn модели!")
print("У всех есть методы fit(), predict(), и т.д.")
```

# Часть 6: Магические методы для ML (10 минут)
## 6.1. Создание удобного класса для данных

```python
class Dataset:
    """Класс для работы с датасетом."""
    
    def __init__(self, data, labels=None):
        self.data = np.array(data)
        self.labels = np.array(labels) if labels is not None else None
        self._current_index = 0
    
    def __len__(self):
        """Возвращает количество样本."""
        return len(self.data)
    
    def __getitem__(self, idx):
        """Позволяет использовать dataset[idx]."""
        if self.labels is not None:
            return self.data[idx], self.labels[idx]
        return self.data[idx]
    
    def __iter__(self):
        """Делает класс итерируемым."""
        self._current_index = 0
        return self
    
    def __next__(self):
        """Для итерации."""
        if self._current_index >= len(self.data):
            raise StopIteration
        result = self[self._current_index]
        self._current_index += 1
        return result
    
    def __repr__(self):
        """Представление для разработчиков."""
        shape = self.data.shape
        return f"Dataset(shape={shape}, has_labels={self.labels is not None})"
    
    def __add__(self, other):
        """Объединение двух датасетов."""
        if isinstance(other, Dataset):
            new_data = np.vstack([self.data, other.data])
            new_labels = None
            if self.labels is not None and other.labels is not None:
                new_labels = np.hstack([self.labels, other.labels])
            return Dataset(new_data, new_labels)
        return NotImplemented

# Демонстрация
# Создаем датасет
X = np.random.randn(10, 3)
y = np.random.randint(0, 2, 10)

dataset = Dataset(X, y)
print(dataset)  # __repr__

print(f"\nРазмер датасета: {len(dataset)}")  # __len__

print("\nПервые 3 элемента:")
for i in range(3):
    data, label = dataset[i]  # __getitem__
    print(f"  {i}: {data} -> {label}")

print("\nИтерация по датасету:")
for i, (data, label) in enumerate(dataset):  # __iter__, __next__
    print(f"  Элемент {i}: метка {label}")
    if i >= 2: break  # Покажем только первые 3

# Объединение датасетов
dataset2 = Dataset(np.random.randn(5, 3), np.random.randint(0, 2, 5))
combined = dataset + dataset2  # __add__
print(f"\nОбъединенный датасет: {combined}, размер: {len(combined)}")
```
