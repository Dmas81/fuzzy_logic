# 1. Введение: Что такое типы данных?

В Python (и программировании в целом) данные бывают разными. Тип данных определяет:

- Какие значения может принимать переменная.
- Какие операции можно над ней выполнять (складывать, делить, соединять и т.д.).

Python — язык с динамической типизацией. Это значит, что нам не нужно явно указывать тип переменной (как в C++ или Java), интерпретатор определяет его сам во время выполнения кода.

# 2. Числовые типы данных
## 2.1. int (Integer) — Целые числа
Это числа без дробной части. Могут быть сколь угодно длинными (в отличие от многих других языков).

```python
# Примеры целых чисел
age = 25
population = 7_900_000_000  # Подчеркивания для удобства чтения (Python 3.6+)
negative_temp = -10

print("Возраст:", age)
print("Население:", population)
print("Тип переменной age:", type(age))
```

Операции с int:

```python
a = 10
b = 3

print("Сложение:", a + b)       # 13
print("Вычитание:", a - b)      # 7
print("Умножение:", a * b)      # 30
print("Деление:", a / b)        # 3.333... (всегда возвращает float!)
print("Целочисленное деление:", a // b)  # 3 (отбрасываем дробную часть)
print("Остаток от деления:", a % b)      # 1
print("Возведение в степень:", a ** b)   # 1000

```

## 2.2. float (Floating Point) — Вещественные числа
Числа с плавающей точкой (десятичная дробь).

```python
# Примеры float
pi = 3.14159
temperature = 36.6
big_float = 2.5e5  # Научная нотация: 2.5 * 10^5 = 250000.0

print("Число Пи:", pi)
print("Тип данных температуры:", type(temperature))
print("Большое число:", big_float)
```

Важная особенность: Арифметика с float может давать небольшие погрешности из-за двоичного представления чисел в памяти компьютера.

```python
# Погрешность вычислений
result = 0.1 + 0.2
print("0.1 + 0.2 =", result)  # Ожидаем 0.3, но видим 0.30000000000000004
print("Это все еще float?", type(result))
```

# 3. Строковый тип данных str (String)

Строка — это последовательность символов (букв, цифр, знаков). Строки в Python неизменяемы (immutable). Создаются с помощью одинарных ' или двойных " кавычек.

```python
# Создание строк
name = "Alice"
profession = 'Data Scientist'
multiline = """Это многострочная
строка. Она удобна для больших текстов."""

print(name)
print(profession)
print(multiline)
```

## 3.1. Основные операции со строками

```python
# Конкатенация (сложение)
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print("Полное имя:", full_name)

# Дублирование (умножение)
laugh = "Ha" * 3
print("Смех:", laugh)  # HaHaHa

# Длина строки (функция len)
print("Длина имени:", len(first_name))
```

## 3.2. Форматирование строк (f-strings)
Начиная с Python 3.6, f-строки — это лучший способ вставлять переменные в текст.

```python
age = 30
city = "Moscow"

# Ставим f перед кавычками и пишем {имя_переменной}
info = f"Меня зовут {full_name}, мне {age} лет, я живу в {city}."
print(info)

# Можно выполнять вычисления прямо внутри {}
print(f"Через 5 лет мне будет {age + 5} лет.")
```

# 4. Булевый тип bool (Boolean)

Используется для логических операций. Принимает только два значения:
True (Истина)
False (Ложь)

```python
is_sunny = True
is_weekend = False

print("Сегодня солнечно?", is_sunny)
print("Тип данных:", type(is_sunny))
```

Операторы сравнения (всегда возвращают bool):

```python
x = 10
y = 20

print("x > y:", x > y)          # False
print("x < y:", x < y)          # True
print("x == y:", x == y)        # False (проверка на равенство)
print("x != y:", x != y)        # True (не равно)
print("x >= 10:", x >= 10)      # True
```

Логические операторы:

```python
has_money = True
is_hungry = False

# and (И) — все условия должны быть True
can_eat = has_money and is_hungry
print("Могу поесть:", can_eat)  # False

# or (ИЛИ) — хотя бы одно условие True
need_to_go_out = has_money or is_hungry
print("Нужно выйти:", need_to_go_out)  # True

# not (НЕ) — инверсия
print("Обратное голоду:", not is_hungry)  # True
```

# № 5. Преобразование типов (Type Casting)

Иногда нам нужно принудительно изменить тип данных. Например, пользователь вводит строку "123", а мы хотим складывать числа.

```python
# Пример: строка в число
num_str = "100"
num_int = int(num_str)  # Преобразуем строку в int
print(f"Было: {num_str} ({type(num_str)}), стало: {num_int} ({type(num_int)})")

# Ошибка: нельзя сложить строку и число напрямую
# print("Сумма: " + 5)  # TypeError!

# Правильно:
print("Сумма: " + str(5))  # Превращаем число в строку

# Преобразование float в int (отбрасывает дробную часть, НЕ округляет!)
price = 15.99
price_int = int(price)
print(f"Цена без копеек: {price_int}")  # 15

# Преобразование в bool
# Пустые значения, 0, None, [] дают False. Все остальное -> True.
print("bool(0):", bool(0))       # False
print("bool(1):", bool(1))       # True
print("bool(''):", bool(""))     # False
print("bool('Hello'):", bool("Hello"))  # True
```

# 9. Составные (коллекции) типы данных

Простые типы (int, float, str, bool) хранят одно значение. Составные типы позволяют хранить коллекции значений. Python имеет четыре основных встроенных типа коллекций:
- Список (list) — упорядоченный, изменяемый
- Кортеж (tuple) — упорядоченный, неизменяемый
- Множество (set) — неупорядоченный, уникальные элементы, изменяемый
- Словарь (dict) — коллекция пар "ключ: значение", изменяемый

## 9.1. Списки (list)

Список — это упорядоченная коллекция, которая может содержать элементы разных типов. Списки изменяемы (можно добавлять, удалять, изменять элементы). Создаются с помощью квадратных скобок [].

```python
# Создание списков
fruits = ["яблоко", "банан", "апельсин"]
mixed_list = [1, "hello", 3.14, True, None]
empty_list = []

print("Фрукты:", fruits)
print("Смешанный список:", mixed_list)
print("Тип данных:", type(fruits))
```

## 9.1.1 Доступ к элементам списка (индексация)

```python
fruits = ["яблоко", "банан", "апельсин", "манго", "киви"]

# Положительные индексы (с начала)
print("Первый элемент:", fruits[0])     # яблоко
print("Второй элемент:", fruits[1])     # банан
print("Последний элемент:", fruits[-1]) # киви (отрицательная индексация)

# Срезы (slicing) — [start:stop:step]
print("Элементы с 1 по 3:", fruits[1:4])     # ['банан', 'апельсин', 'манго']
print("Первые три элемента:", fruits[:3])    # ['яблоко', 'банан', 'апельсин']
print("Каждый второй:", fruits[::2])         # ['яблоко', 'апельсин', 'киви']
print("В обратном порядке:", fruits[::-1])   # ['киви', 'манго', 'апельсин', 'банан', 'яблоко']
```

## 9.1.2 Изменение списков

```python
numbers = [1, 2, 3, 4, 5]
print("Исходный список:", numbers)

# Изменение элемента
numbers[0] = 100
print("После изменения первого элемента:", numbers)

# Добавление элементов
numbers.append(6)              # Добавить в конец
print("После append:", numbers)

numbers.insert(2, 999)         # Вставить на позицию 2
print("После insert:", numbers)

# Удаление элементов
numbers.remove(999)            # Удалить по значению
print("После remove:", numbers)

popped = numbers.pop()         # Удалить и вернуть последний элемент
print(f"Удален элемент: {popped}, список теперь: {numbers}")

numbers.pop(0)                 # Удалить по индексу
print("После pop(0):", numbers)

# Расширение списка
more_numbers = [10, 20, 30]
numbers.extend(more_numbers)
print("После extend:", numbers)
```

## 9.1.3 Полезные методы и операции

```python
import copy

fruits = ["яблоко", "банан", "апельсин", "банан", "киви"]

# Поиск и подсчет
print("Длина списка:", len(fruits))
print("Индекс 'апельсин':", fruits.index("апельсин"))
print("Количество 'банан':", fruits.count("банан"))

# Сортировка
fruits.sort()                   # Сортировка на месте
print("После sort():", fruits)

fruits.reverse()                # Разворот списка
print("После reverse():", fruits)

# Проверка наличия
print("Есть ли 'манго'?", "манго" in fruits)    # False
print("Есть ли 'яблоко'?", "яблоко" in fruits)  # True

# Копирование (важно!)
original = [1, 2, [3, 4]]
shallow_copy = original.copy()     # Поверхностная копия
deep_copy = copy.deepcopy(original) # Глубокая копия (нужен import copy)
# print(id(original), id(shallow_copy), id(deep_copy))

# List comprehension (генератор списков) — мощный инструмент
squares = [x**2 for x in range(10)]
print("Квадраты чисел 0-9:", squares)

even_numbers = [x for x in range(20) if x % 2 == 0]
print("Четные числа до 20:", even_numbers)
```

## 9.2. Кортежи (tuple)

Кортеж — это упорядоченная коллекция, похожая на список, но неизменяемая (immutable). Создается с помощью круглых скобок ().

```python
# Создание кортежей
coordinates = (10, 20)
rgb = (255, 128, 0)
single_element = (5,)          # Кортеж из одного элемента (важна запятая!)
empty_tuple = ()

print("Координаты:", coordinates)
print("Тип:", type(coordinates))
```

## 9.2.1 Операции с кортежами

```python
point = (5, 10, 15)

# Доступ к элементам (как в списке)
print("x:", point[0])          # 5
print("y:", point[1])          # 10
print("Последний:", point[-1]) # 15

# Срезы работают
print("Первые два:", point[:2]) # (5, 10)

# Кортежи неизменяемы
try:
    point[0] = 100
except TypeError as e:
    print(f"Ошибка: {e}")  # 'tuple' object does not support item assignment

# Распаковка кортежа (очень удобно!)
person = ("Анна", 25, "Москва")
name, age, city = person
print(f"Имя: {name}, Возраст: {age}, Город: {city}")

# Обмен значениями через распаковку
(a, b) = (5, 10)
print(f"До обмена: a={a}, b={b}")
(a, b) = (b, a)
print(f"После обмена: a={a}, b={b}")
```

## 9.2.2 Когда использовать кортежи?

```python
# 1. Географические координаты (не должны меняться)
moscow_coords = (55.7558, 37.6173)

# 2. Возврат нескольких значений из функции
def get_min_max(numbers):
    return min(numbers), max(numbers)

result = get_min_max([3, 1, 8, 4, 9, 2])
print("Min и Max:", result)  # (1, 9)

# 3. Ключи в словарях (списки нельзя использовать как ключи)
# dict_with_list = {[1,2]: "value"}  # Ошибка!
dict_with_tuple = {(1,2): "value"}    # Работает
print("Словарь с кортежем-ключом:", dict_with_tuple)
```

## 9.3 Множества (set)

Множество — это неупорядоченная коллекция уникальных элементов. Изменяемо (можно добавлять/удалять элементы), но сами элементы должны быть неизменяемыми. Создается с помощью фигурных скобок {} или функции set().

```python
# Создание множеств
colors = {"красный", "синий", "зеленый", "красный"}  # Дубликат удалится
print("Множество цветов:", colors)  # {'синий', 'зеленый', 'красный'}

# Пустое множество (важно!)
empty_set = set()          # {} — это пустой словарь, а не множество!
print("Пустое множество:", empty_set)
print("Тип:", type(empty_set))
```

## 9.3.1 Операции с множествами

```python
# Добавление и удаление
fruits_set = {"яблоко", "банан"}
fruits_set.add("апельсин")
print("После add:", fruits_set)

fruits_set.update(["манго", "киви"])  # Добавить несколько элементов
print("После update:", fruits_set)

fruits_set.remove("банан")    # Удалить (ошибка если нет)
print("После remove:", fruits_set)

fruits_set.discard("яблоко")  # Удалить без ошибки если нет
fruits_set.discard("несуществующий")  # Не вызовет ошибку

popped = fruits_set.pop()     # Удалить и вернуть случайный элемент
print(f"Удален случайный элемент: {popped}, осталось: {fruits_set}")

# Проверка наличия
print("Есть ли 'манго'?", "манго" in fruits_set)
```

## 9.3.2 Математические операции с множествами

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

# Объединение (union)
union = set_a | set_b  # или set_a.union(set_b)
print("Объединение:", union)  # {1, 2, 3, 4, 5, 6, 7, 8}

# Пересечение (intersection)
intersection = set_a & set_b  # или set_a.intersection(set_b)
print("Пересечение:", intersection)  # {4, 5}

# Разность (difference)
difference = set_a - set_b  # или set_a.difference(set_b)
print("Разность A - B:", difference)  # {1, 2, 3}

# Симметрическая разность (symmetric difference)
sym_diff = set_a ^ set_b  # или set_a.symmetric_difference(set_b)
print("Симметрическая разность:", sym_diff)  # {1, 2, 3, 6, 7, 8}

# Проверка подмножества
set_c = {1, 2}
print("C ⊆ A?", set_c.issubset(set_a))  # True
print("A ⊇ C?", set_a.issuperset(set_c))  # True
```

## 9.3.3 Практическое применение множеств

```python
# Удаление дубликатов из списка
numbers_with_duplicates = [1, 2, 3, 2, 1, 4, 5, 3, 2]
unique_numbers = list(set(numbers_with_duplicates))
print("Уникальные числа:", unique_numbers)  # [1, 2, 3, 4, 5] (порядок может быть любым)

# Поиск общих элементов
group1 = {"Анна", "Иван", "Петр", "Мария"}
group2 = {"Иван", "Ольга", "Петр", "Сергей"}

common_members = group1 & group2
print("Общие участники:", common_members)  # {'Иван', 'Петр'}

# Проверка уникальности
def has_duplicates(items):
    return len(items) != len(set(items))

print("Есть дубликаты?", has_duplicates([1, 2, 3, 4, 5]))     # False
print("Есть дубликаты?", has_duplicates([1, 2, 3, 2, 5]))     # True
```

## 9.4. Словари (dict)

Словарь — это коллекция пар "ключ: значение". Ключи должны быть уникальными и неизменяемыми (строки, числа, кортежи). Значения могут быть любыми. Словари изменяемы и очень эффективны для поиска по ключу.

```python
# Создание словарей
person = {
    "name": "Анна",
    "age": 28,
    "city": "Москва",
    "is_student": False
}

# Альтернативные способы
empty_dict = {}
person2 = dict(name="Иван", age=30, city="СПб")

print("Словарь person:", person)
print("Тип:", type(person))
```

## 9.4.1. Доступ к значениям

```python
person = {"name": "Анна", "age": 28, "city": "Москва"}

# Доступ по ключу
print("Имя:", person["name"])
print("Возраст:", person["age"])

# Безопасный доступ (возвращает None или значение по умолчанию)
print("Город:", person.get("city"))           # Москва
print("Работа:", person.get("job"))            # None (нет ошибки)
print("Работа:", person.get("job", "не указана"))  # не указана

# Ошибка при отсутствии ключа
try:
    print(person["job"])  # KeyError!
except KeyError as e:
    print(f"Ошибка: ключ {e} не найден")
```

## 9.4.2 Изменение и добавление элементов

Создайте программу для управления списком контактов. Контакт должен содержать: имя, телефон, email, теги (список строк).

Требования:
- Хранить контакты в словаре, где ключ — имя контакта
- Реализовать функции:
- Добавление контакта
- Поиск по тегу
- Удаление контакта

Использовать различные типы данных в комбинации

```python
# Решение (базовое)
contacts = {}  # Словарь для хранения контактов

def add_contact(name, phone, email, tags=None):
    """Добавление нового контакта"""
    if tags is None:
        tags = []

    contacts[name] = {
        "phone": phone,
        "email": email,
        "tags": tags
    }
    print(f"Контакт {name} добавлен!")

def find_by_tag(tag):
    """Поиск контактов по тегу"""
    result = []
    for name, info in contacts.items():
        if tag in info["tags"]:
            result.append(name)
    return result

def delete_contact(name):
    """Удаление контакта"""
    if name in contacts:
        del contacts[name]
        print(f"Контакт {name} удален")
        return True
    else:
        print(f"Контакт {name} не найден")
        return False

def show_all_contacts():
    """Показать все контакты"""
    if not contacts:
        print("Список контактов пуст")
        return

    print("\n=== Список контактов ===")
    for name, info in contacts.items():
        print(f"\nИмя: {name}")
        print(f"  Телефон: {info['phone']}")
        print(f"  Email: {info['email']}")
        print(f"  Теги: {', '.join(info['tags']) if info['tags'] else 'нет'}")

# Пример использования
add_contact("Анна", "+7-999-123-4567", "anna@example.com", ["работа", "коллега"])
add_contact("Иван", "+7-999-765-4321", "ivan@example.com", ["друзья", "учеба"])
add_contact("Мария", "+7-999-111-2233", "maria@example.com", ["работа"])

show_all_contacts()

print("\nПоиск по тегу 'работа':", find_by_tag("работа"))
print("Поиск по тегу 'друзья':", find_by_tag("друзья"))

delete_contact("Иван")
show_all_contacts()
```

# 4. Кортежи: неизменяемые структуры для ML
