# Часть 2: Основные типы данных в Python (60 минут)
Переходим к фундаменту. Всё в Python является объектом, и у каждого объекта есть свой тип.

## 2.1. Числа: int и float
(Демонстрация в новых ячейках)

```python
# Целые числа (int)
a = 10
b = -5
c = 1234567890

print(a, b, c)
print("Тип переменной a:", type(a))
```

```python
# Числа с плавающей точкой (float)
pi = 3.14159
temp = -2.5
science = 5.97e24  # Научная нотация (масса Земли в кг)

print(pi, temp, science)
print("Тип переменной pi:", type(pi))
```

```python
# Арифметика с разными типами
x = 10  # int
y = 3.0 # float

result = x + y
print(result, type(result)) # Результат будет float. Python автоматически приводит типы к более "широкому".
```

Важно для ML: В машинном обучении мы постоянно работаем с вещественными числами (float32, float64), так как веса моделей и признаки — это обычно не целые числа.

## 2.2. Строки: str

Строки используются для текстовых данных. В анализе данных это могут быть названия, комментарии, целые документы (для NLP).

```python
# Способы создания строк
name = "Алиса"
surname = 'Петрова'
message = """Это многострочный
текст, который можно использовать
для длинных описаний."""

print(name)
print(surname)
print(message)
```

```python
# Основные операции со строками
first_name = "Иван"
last_name = "Иванов"

# Конкатенация (склеивание)
full_name = first_name + " " + last_name
print(full_name)

# Дублирование
print("ha" * 3)

# Длина строки
print("Длина строки full_name:", len(full_name))

# Доступ по индексу (строка - это последовательность символов)
print("Первый символ:", full_name[0])
print("Последний символ:", full_name[-1])

# Срезы (slicing) - важнейший инструмент!
print("Первые 4 символа:", full_name[:4])
print("Символы с 6-го по 9-й:", full_name[5:9])
```

```python
# Полезные методы строк (некоторые варианты)
text = "  Привет, Мир!  "
print(text.upper())
print(text.lower())
print(text.strip()) # Удаляет пробелы в начале и конце
print(text.replace("Мир", "Python"))
```

## 2.3. Булев тип: bool

Булевы значения (True/False) — это результат логических операций. Они повсюду в программировании: условия циклов, фильтрация данных.

```python
is_sunny = True
is_rainy = False

print(is_sunny, type(is_sunny))
```

```python
# Операции сравнения
x = 10
y = 5

print("x > y :", x > y)
print("x < y :", x < y)
print("x == y:", x == y)
print("x != y:", x != y)
print("x >= 10:", x >= 10)
```

```python
# Логические операции (and, or, not)
a = True
b = False

print("a and b:", a and b) # Истина, только если оба истинны
print("a or b:", a or b)   # Истина, если хотя бы один истинен
print("not a:", not a)     # Инвертирует значение
```

## 2.4. Специальный тип None

None обозначает "отсутствие значения". Это аналог null в других языках. Очень часто используется для обозначения того, что переменная или результат функции еще не определены.

```python
result = None
print(result, type(result))

# Проверка на None
if result is None:
    print("Результата пока нет!")
```

# Часть 3: Базовые структуры данных (Collections) (40 минут)
Для хранения и обработки наборов данных нам нужны контейнеры.

## 3.1. Списки: list

Список — это упорядоченная, изменяемая коллекция элементов. Самый популярный тип для хранения последовательностей данных.

```python
# Создание списков
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True, None]  # Может содержать разные типы
empty = []

print(numbers)
print(mixed)
print(empty)
```

```python
# Доступ к элементам и срезы (как в строках)
fruits = ["apple", "banana", "cherry", "date"]
print(fruits[0])      # apple
print(fruits[-1])     # date
print(fruits[1:3])    # ['banana', 'cherry'] (срез, последний индекс не включается)
```

```python
# Списки изменяемы
fruits[1] = "blueberry"
print(fruits)         # ['apple', 'blueberry', 'cherry', 'date']
```

```python
# Основные методы списков
stack = []
stack.append(1)        # Добавить в конец
stack.append(2)
stack.append(3)
print("После append:", stack)

last = stack.pop()     # Извлечь с конца
print("Извлекли:", last)
print("После pop:", stack)

numbers = [5, 2, 8, 1, 9]
numbers.sort()         # Сортировка на месте
print("Отсортировано:", numbers)

print("Длина списка:", len(numbers))
print("Индекс числа 8:", numbers.index(8))
```

## 3.2. Кортежи: tuple

Кортеж — это упорядоченная, неизменяемая коллекция. Используется для защиты данных от изменений и может быть ключом в словаре.

```python
# Создание кортежа
point = (10, 20)
rgb_color = (255, 128, 0)
single_element = (42,)  # Запятая обязательна для кортежа из одного элемента

print(point, type(point))
print(point[0])  # Доступ по индексу есть
# point[0] = 100  # Попытка изменить вызовет ошибку TypeError
```

## 3.3. Словари: dict

Словарь хранит пары "ключ-значение". Ключи должны быть уникальными и неизменяемыми (строки, числа, кортежи). Это аналог структуры или ассоциативного массива. В ML словари часто используются для хранения гиперпараметров моделей или помеченных данных.

```python
# Создание словаря
person = {
    "name": "Алиса",
    "age": 25,
    "city": "Москва",
    "is_student": False
}

print(person)
print(type(person))
```

```python
# Доступ к значениям
print(person["name"])
print(person.get("age"))  # Более безопасный способ, вернет None, если ключа нет
# print(person["salary"]) # Ошибка! Такого ключа нет.
print(person.get("salary", "Ключ не найден")) # Можно задать значение по умолчанию
```

```python
# Изменение и добавление элементов
person["age"] = 26            # Изменение существующего
person["job"] = "Data Analyst" # Добавление нового
print(person)
```

```python
# Перебор словаря
for key in person:
    print(f"Ключ: {key}, Значение: {person[key]}")

# Более "питоничный" способ
for key, value in person.items():
    print(f"{key}: {value}")
```

## 3.4. Множества: set

Множество — это неупорядоченная коллекция уникальных элементов. Полезно для удаления дубликатов и проверки принадлежности.

```python
ids = [1, 2, 3, 1, 2, 4, 5, 1]
unique_ids = set(ids)
print(unique_ids)  # {1, 2, 3, 4, 5} (порядок может отличаться)

# Проверка принадлежности
print(3 in unique_ids)  # True
print(10 in unique_ids) # False
```

## 4.1. Введение в NumPy

NumPy — это основа для научных вычислений. Его главный объект — многомерный массив (ndarray). Все данные в ML в итоге превращаются в такие массивы.

```python
pip install numpy
```

```python
import numpy as np  # Стандартный псевдоним

# Создаем массив из обычного списка
list_data = [1, 2, 3, 4, 5]
np_array = np.array(list_data)
print("NumPy массив:", np_array)
print("Тип элементов:", np_array.dtype)
print("Размерность:", np_array.shape)
```

```python
# Умножение массива на число (позволяет избежать циклов!)
print("Исходный массив:", np_array)
print("Умноженный на 2:", np_array * 2)
print("Синус каждого элемента:", np.sin(np_array))
```

# 4.2. Введение в Pandas

Pandas работает с таблицами (DataFrame), как Excel, но мощнее. Строится на основе NumPy.

```python
pip install pandas
```

```python
import pandas as pd

# Создаем DataFrame из словаря
data = {
    'Name': ['Алиса', 'Борис', 'Виктор'],
    'Age': [25, 30, 35],
    'Salary': [70000, 80000, 90000]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Простейший анализ
print("\nСредний возраст:", df['Age'].mean())
print("Максимальная зарплата:", df['Salary'].max())
print("\nПервые 2 строки:")
print(df.head(2))
```
