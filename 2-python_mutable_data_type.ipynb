# Часть 1: Введение в концепцию мутабельности (10 минут)
## 1.1. Что такое "изменяемость"?

# Изменяемые (mutable) и неизменяемые (immutable) объекты

В Python всё является объектом. У каждого объекта есть:
1. **Идентификатор (id)** - уникальный номер объекта в памяти.
2. **Тип (type)** - int, str, list и т.д.
3. **Значение (value)** - собственно данные.

**Мутабельность** - это способность объекта изменять свое значение после создания, сохраняя при этом свой идентификатор (оставаясь тем же объектом в памяти).

## 1.2. Функции id() и is

Прежде чем говорить об изменяемости, научимся "заглядывать" в память Python.

```python
# Функция id() показывает уникальный идентификатор объекта в памяти
a = 5
b = 5

print("ID переменной a:", id(a))
print("ID переменной b:", id(b))

# Интересно! Они могут иметь одинаковый id (интернирование малых целых чисел)
print("a is b?", a is b)  # Оператор is сравнивает идентификаторы!

# Но это не всегда так
c = 1000
d = 1000
print("\nID переменной c:", id(c))
print("ID переменной d:", id(d))
print("c is d?", c is d)  # Скорее всего False (зависит от реализации)
print("c == d?", c == d)  # == сравнивает значения!
```

Объясняем: is проверяет, указывают ли переменные на один и тот же объект в памяти. == проверяет, равны ли значения объектов.

# Часть 2: Неизменяемые типы данных (Immutable) (25 минут)
## 2.1. Числа (int, float)

```python
# Числа неизменяемы
x = 10
print(f"x = {x}, id(x) = {id(x)}")

x = x + 5  # Создается НОВЫЙ объект!
print(f"x = {x}, id(x) = {id(x)}")  # id изменился!

# Проверим нагляднее
print("\nПошагово:")
y = 20
print(f"id(y) до: {id(y)}")
y += 1  # То же самое что y = y + 1
print(f"id(y) после y += 1: {id(y)}")  # id другой!
```

Ключевой момент: Когда мы "изменяем" число, на самом деле создается новый объект, а переменная начинает ссылаться на него. Старый объект (если на него никто не ссылается) будет удален сборщиком мусора.

## 2.2. Строки (str)

Строки - классический пример неизменяемого типа. Многие методы строк создают новые строки, не изменяя исходную.

```python
# Строки неизменяемы
text = "Hello"
print(f"text = '{text}', id(text) = {id(text)}")

# Попробуем изменить символ
# text[0] = 'J'  # ОШИБКА! TypeError: 'str' object does not support item assignment

# Методы строк возвращают НОВЫЕ строки
new_text = text.upper()
print(f"После upper(): '{new_text}', id(new_text) = {id(new_text)}")
print(f"Исходная строка: '{text}', id(text) = {id(text)}")  #не изменилась!

# Конкатенация создает новую строку
text2 = text + " World"
print(f"После конкатенации: '{text2}', id(text2) = {id(text2)}")
print(f"Исходная строка все еще: '{text}', id = {id(text)}")
```

```python
# Демонстрация с циклом (почему это важно)
# НЕ ЭФФЕКТИВНО! (каждый раз создается новая строка)
result = ""
for i in range(5):
    result = result + str(i)
    print(f"Итерация {i}: '{result}', id = {id(result)}")

# Эффективный способ: использовать список и join
parts = []
for i in range(5):
    parts.append(str(i))
    print(f"Список частей: {parts}, id списка = {id(parts)}")  # id списка не меняется!
    
result = "".join(parts)
print(f"Результат join: '{result}'")
```

## 2.3. Кортежи (tuple)

Кортежи - неизменяемые "списки".

```python
# Кортежи неизменяемы
t = (1, 2, 3)
print(f"Кортеж t = {t}, id(t) = {id(t)}")

# t[0] = 10  # ОШИБКА! Кортеж неизменяем

# Но внимание! Если кортеж содержит изменяемые объекты...
t2 = ([1, 2], 3, 4)
print(f"Кортеж со списком: {t2}")
print(f"id кортежа t2: {id(t2)}")

# Мы НЕ можем изменить сам кортеж
# t2[0] = [10, 20]  # ОШИБКА!

# Но МОЖЕМ изменить содержимое списка внутри кортежа!
t2[0].append(100)
print(f"После изменения списка внутри кортежа: {t2}")
print(f"id кортежа остался тем же: {id(t2)}")  # id не изменился!

print("\nВывод: кортеж неизменяем только в отношении своей структуры")
print("(какие объекты на каких позициях лежат), но не обязательно")
print("в отношении самих этих объектов, если они изменяемы.")
```

## 2.4. Проверка неизменяемости через функции

```python
def try_to_modify(x):
    """Демонстрация передачи неизменяемого объекта в функцию"""
    print(f"Внутри функции ДО: x = {x}, id(x) = {id(x)}")
    x = x * 2  # Создаем новый объект
    print(f"Внутри функции ПОСЛЕ: x = {x}, id(x) = {id(x)}")
    return x

value = 5
print(f"Снаружи ДО вызова: value = {value}, id(value) = {id(value)}")
result = try_to_modify(value)
print(f"Снаружи ПОСЛЕ вызова: value = {value}, id(value) = {id(value)}")
print(f"Результат функции: {result}, id(result) = {id(result)}")
print(f"value не изменилось! Функция работала с копией ссылки на объект.")
```

# Часть 3: Изменяемые типы данных (Mutable) (25 минут)
## 3.1. Списки (list)

Списки - самый популярный изменяемый тип.

```python
# Списки изменяемы
lst = [1, 2, 3]
print(f"Список lst = {lst}, id(lst) = {id(lst)}")

# Изменяем элемент
lst[0] = 100
print(f"После lst[0] = 100: {lst}, id(lst) = {id(lst)}")  # id тот же!

# Добавляем элемент
lst.append(4)
print(f"После append(4): {lst}, id(lst) = {id(lst)}")  # id тот же!

# Удаляем элемент
popped = lst.pop()
print(f"После pop(): {lst}, id(lst) = {id(lst)}")  # id тот же!
```

## 3.2. Словари (dict)

```python
# Словари изменяемы
d = {"name": "Алиса", "age": 25}
print(f"Словарь d = {d}, id(d) = {id(d)}")

# Изменяем значение
d["age"] = 26
print(f"После d['age'] = 26: {d}, id(d) = {id(d)}")  # id тот же!

# Добавляем новую пару
d["city"] = "Москва"
print(f"После добавления city: {d}, id(d) = {id(d)}")  # id тот же!

# Удаляем
del d["name"]
print(f"После удаления name: {d}, id(d) = {id(d)}")  # id тот же!
```

## 3.3. Множества (set)

```python
# Множества изменяемы
s = {1, 2, 3}
print(f"Множество s = {s}, id(s) = {id(s)}")

s.add(4)
print(f"После add(4): {s}, id(s) = {id(s)}")  # id тот же!

s.remove(2)
print(f"После remove(2): {s}, id(s) = {id(s)}")  # id тот же!
```

# Часть 4: Важные последствия мутабельности (20 минут)
## 4.1. Копирование: поверхностное и глубокое

Это одна из самых частых причин багов у начинающих!

```python
# Проблема с изменяемыми объектами
list_a = [1, 2, [3, 4]]
list_b = list_a  # Это НЕ копия! Это новый ярлык на тот же объект

print("До изменений:")
print(f"list_a = {list_a}, id = {id(list_a)}")
print(f"list_b = {list_b}, id = {id(list_b)}")

# Изменяем через list_b
list_b[0] = 100
list_b[2].append(5)

print("\nПосле изменений через list_b:")
print(f"list_a = {list_a}")  # Ой! list_a тоже изменился!
print(f"list_b = {list_b}")
print(f"id list_a: {id(list_a)}")
print(f"id list_b: {id(list_b)}")
```

```python
# Решение: создаем поверхностную копию
import copy

list_a = [1, 2, [3, 4]]
list_c = list_a.copy()  # или list_a[:]

print("Поверхностная копия:")
print(f"list_a = {list_a}, id = {id(list_a)}")
print(f"list_c = {list_c}, id = {id(list_c)}")

# Изменяем через list_c
list_c[0] = 200
list_c[2].append(6)

print("\nПосле изменений через list_c:")
print(f"list_a = {list_a}")  # Внешний элемент не изменился, но внутренний список...
print(f"list_c = {list_c}")
# Внутренний список [3, 4] все еще общий для обоих!
```

```python
# Решение: глубокая копия
list_a = [1, 2, [3, 4]]
list_d = copy.deepcopy(list_a)

print("Глубокая копия:")
print(f"list_a = {list_a}, id = {id(list_a)}")
print(f"list_d = {list_d}, id = {id(list_d)}")

# Изменяем через list_d
list_d[0] = 300
list_d[2].append(7)

print("\nПосле изменений через list_d:")
print(f"list_a = {list_a}")  # Полностью независим!
print(f"list_d = {list_d}")
```

# 4.2. Изменяемые объекты как аргументы по умолчанию

Классическая "ловушка" Python!

```python
# ПЛОХОЙ пример
def add_item_bad(item, my_list=[]):
    """Добавляет элемент в список. С проблемой!"""
    my_list.append(item)
    return my_list

print(add_item_bad(1))  # [1]
print(add_item_bad(2))  # [1, 2] - Ой! А ожидали [2]
print(add_item_bad(3))  # [1, 2, 3] - Список продолжает расти!

# Проверим, в чем дело
print(f"Функция add_item_bad, ее __defaults__: {add_item_bad.__defaults__}")
# Список по умолчанию создается ОДИН РАЗ при определении функции!
```

```python
# ХОРОШИЙ пример
def add_item_good(item, my_list=None):
    """Добавляет элемент в список. Без проблем."""
    if my_list is None:
        my_list = []  # Создаем новый список при каждом вызове
    my_list.append(item)
    return my_list

print(add_item_good(1),id(add_item_good))        # [1]
print(add_item_good(2),id(add_item_good))        # [2] - работает как ожидалось!
print(add_item_good(3, [10]), id(add_item_good))  # [10, 3] - и можно передать свой список
```

# 4.3. Практический пример: обработка данных

Показываем, как понимание мутабельности помогает в реальной задаче.

```python
# Представим, что мы обрабатываем данные о студентах
students = [
    {"name": "Алиса", "grades": [85, 90, 78]},
    {"name": "Борис", "grades": [92, 88, 95]},
    {"name": "Виктор", "grades": [70, 85, 80]}
]

def add_bonus_all(students_list, bonus=5):
    """Добавляет бонус ко всем оценкам всех студентов."""
    for student in students_list:
        # Мы изменяем содержимое списков grades, но не сами списки студентов
        grades = student["grades"]
        for i in range(len(grades)):
            grades[i] = grades[i] + bonus
    # Мы НЕ создаем новый список студентов, а изменяем существующий
    return students_list

print("До добавления бонуса:")
for s in students:
    print(f"{s['name']}: {s['grades']}")

# Применяем функцию
result = add_bonus_all(students)

print("\nПосле добавления бонуса:")
for s in students:
    print(f"{s['name']}: {s['grades']}")
    
print(f"\nresult is students? {result is students}")  # True - это тот же объект!
```

```python
# А если мы хотим сохранить исходные данные?
import copy

students_original = [
    {"name": "Алиса", "grades": [85, 90, 78]},
    {"name": "Борис", "grades": [92, 88, 95]},
]

def add_bonus_safe(students_list, bonus=5):
    """Возвращает НОВЫЙ список студентов с бонусом, не изменяя исходный."""
    students_new = copy.deepcopy(students_list)  # Создаем полную независимую копию
    
    for student in students_new:
        grades = student["grades"]
        for i in range(len(grades)):
            grades[i] = grades[i] + bonus
            
    return students_new

students_with_bonus = add_bonus_safe(students_original)

print("Исходные студенты (не изменились):")
for s in students_original:
    print(f"{s['name']}: {s['grades']}")

print("\nСтуденты с бонусом (новая копия):")
for s in students_with_bonus:
    print(f"{s['name']}: {s['grades']}")
```
